# Python problems

1. Count numbers greater than the average

Given a list of numbers, calculate the average and return how many numbers are greater than it.


```python
def greater(number):
  tot = sum(number)
  avg = tot/len(number)
  count = sum (1 for i in number if i > avg)
  return count

numbers = [5, 10, 15, 20, 25]
greater(numbers)
```

3. Find the second-largest number

Find the second-largest unique number without using sort() or sorted()

numbers = [10, 5, 20, 8, 20, 15]

Expected output: 15

```python
numbers = [10, 5, 20, 8, 20, 15]

max_ = 0
sec_max_ = 0
for i in range(len(numbers)):
  if numbers[i] > max_:
    max_ = numbers[i]
    for j in range(i+1, (len(numbers))):
      if numbers[j] < max_ and numbers[j] > sec_max_:
        sec_max_ = numbers[j] 

print(max_)
print(sec_max_)
```
